```python
# 1768. Merge Strings Alternately

word1 = 'abc'
word2 = 'pqr'
merge = ''
for i in word1:
    for j in word2:
        merge+=i+j
print(merge)
```

```python
s = "a"
t = "aa"

for i in t:
    if i not in s:
        print(i)
    
```

```python
nums1 = [1,2]
nums2 = [3,4]

nl = nums1 + nums2
snl = sorted(nl)
n = len(snl)
if n %2 == 0:
    median = (snl[(n//2)-1] + snl[(n//2)])/2
    print(float(median))
else:
    median = snl[(n//2)]
    print(float(median))
```

```python
l = [1,3,5,2,1]
n = len(l)
for i in range(n-1):
    for j in range(n-i-1):
        if l[j]>l[j+1]:
            l[j],l[j+1]=l[j+1],l[j]
print(l)

```
