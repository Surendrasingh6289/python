```python
from ucimlrepo import fetch_ucirepo 
bank_marketing = fetch_ucirepo(id=222) 

# data (as pandas dataframes) 
x = bank_marketing.data.features 
y = bank_marketing.data.targets 

# metadata 
print(bank_marketing.metadata) 

# variable information 
print(bank_marketing.variables) 
```

```python
x.info()
```

```python
x.isnull().sum()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
x.drop_duplicates()
```

```python
x.drop(columns=['contact','poutcome'],inplace=True)
```

```python
x['education'].fillna(x['education'].mode()[0],inplace=True)
x['job'].fillna(x['job'].mode()[0],inplace=True)
```

```python
x
```

```python
object_col = x.select_dtypes(include='object')
object_col
```

```python
from sklearn.preprocessing import LabelEncoder,MinMaxScaler
import joblib
encoder = {}
for i in object_col:
    lab_enc = LabelEncoder()
    x[i] = lab_enc.fit_transform(x[i])
    encoder[i] = lab_enc
```

```python
x
```

```python
min_max_scale = MinMaxScaler()
normalized_col = ['age','balance','duration']

x[normalized_col] = min_max_scale.fit_transform(x[normalized_col])
```

```python
y.values.reshape(-1,)
```

```python
y = lab_enc.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3)
```

```python
from sklearn.model_selection import  GridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
paramgrid = {
    'n_estimators':[100,150,200],
    'max_depth':[4,5,8,3],
}
```

```python
rf_model = RandomForestClassifier()
```

```python
grid_model = GridSearchCV(estimator=rf_model,
                        param_grid= paramgrid,
                        cv= 5)
```

```python
grid_model.fit(x_train,y_train)
```

```python
grid_model.best_params_
```

```python
y_pred = grid_model.predict(x_test)
y_pred[:10]
```

```python
y_test[:10]
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
accuracy_score(y_test,y_pred)*100
```

```python
confusion_matrix(y_test,y_pred)
```

```python
joblib.dump(grid_model,"rf_churn_model.pkl")
```
