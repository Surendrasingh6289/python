```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
titanic = pd.read_csv(r"D:\Grow Tech\Datasets\Titanic-Dataset.csv")
titanic
```

```python
titanic.head()
```

```python
titanic.info()
```

```python
titanic.describe()
```

```python
titanic.isnull().sum()
```

```python
titanic.drop_duplicates()
```

```python
titanic['Age'].fillna(titanic['Age'].median(),inplace=True)
```

```python
titanic.isnull().sum()
```

```python
titanic['Embarked'].dropna(inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler
lab_enc = LabelEncoder()
```

```python
for i in titanic.select_dtypes('object'):
    titanic[i] = lab_enc.fit_transform(titanic[i])
titanic
```

```python
X = titanic[['Pclass','Sex', 'Age', 'SibSp', 'Parch', 'Embarked']]
y = titanic['Survived']
```

```python
X,y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
model = GaussianNB()
model.fit(X_train,y_train)
```

```python
multi_model = MultinomialNB()
multi_model.fit(X_train,y_train)
```

```python
ber_model = BernoulliNB()
ber_model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred[:10]
```

```python
y_test[:10]
```

```python
model.score(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,precision_score,recall_score,f1_score,classification_report
```

```python
cm = confusion_matrix(y_test,y_pred)
sns.heatmap(cm,annot=True,fmt="d",cmap="Blues")
plt.show()
```

```python
print(f"Accuracy score: {accuracy_score(y_test,y_pred)*100}")
print(f"Precision score: {precision_score(y_test,y_pred)*100}")
print(f"Recall score: {recall_score(y_test,y_pred)*100}")
print(f"Classfication Report: \n{classification_report(y_test,y_pred)}")
```
