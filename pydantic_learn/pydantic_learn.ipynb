```python
from pydantic import BaseModel,EmailStr
class Student(BaseModel):
    Name: str
    Gender: str
    Age: int
    Location: str

st = Student(Name= 'Chiraag',Gender='Male',Age=33,Location='Indore')
print(st)
```

```python
class Student(BaseModel):
    Name: str
    Gender: str
    Age: int
    Location: str

st = Student(Name= 'Chiraag',Gender='Male',Age='33',Location='Indore')
print(st)
st.model_dump()
```

```python
class Employee(BaseModel):
    Name:str
    Age:int
    Salary:float

data = {
    'Name': "John",
    'Age': 22,
    "Salary": 100000
}

emp = Employee(**data)
emp
```

```python
class Student(BaseModel):
    Name: str
    Gender: str
    Age: int
    Location: str
    email: EmailStr

st = Student(Name= 'Chiraag',Gender='Male',Age='33',Location='Indore',email='sdfghr@gmail.com')
print(st)
st.model_dump()
```
